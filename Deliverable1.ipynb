```python
import pandas as pd
import matplotlib.pyplot as plt
from openpyxl import Workbook
from openpyxl.drawing.image import Image

# Load the Excel file and specify the correct sheet name
file_path = 'GPPS_2023_Practice_results_(weighted)_(Excel)_PUBLIC.xlsx'
sheet_name = 'Overall experience'

# Load the specified sheet and extract the data while skipping unnecessary rows
df = pd.read_excel(file_path, sheet_name=sheet_name, skiprows=9, usecols='A,L,P:U')

# Rename the columns according to the desired order
df.columns = ['Region Name', '% Very good', '% Fairly good', '% Neither good nor poor', '% Fairly poor', '% Very poor']

# Drop rows with missing values
df.dropna(inplace=True)

# Convert percentage columns to numeric types
percentage_columns = ['% Very good', '% Fairly good', '% Neither good nor poor', '% Fairly poor', '% Very poor']
df[percentage_columns] = df[percentage_columns].apply(pd.to_numeric, errors='coerce')

# Set 'Region Name' as the index
df.set_index('Region Name', inplace=True)

# Create a bar chart
plt.figure(figsize=(18, 12))
plt.bar(df.index, df['% Very good'], label='% Very good', color='navy')
plt.bar(df.index, df['% Fairly good'], bottom=df['% Very good'], label='% Fairly good', color='royalblue')
plt.bar(df.index, df['% Neither good nor poor'], bottom=df['% Very good'] + df['% Fairly good'], label='% Neither good nor poor', color='skyblue')
plt.bar(df.index, df['% Fairly poor'], bottom=df['% Very good'] + df['% Fairly good'] + df['% Neither good nor poor'], label='% Fairly poor', color='lightblue')
plt.bar(df.index, df['% Very poor'], bottom=df['% Very good'] + df['% Fairly good'] + df['% Neither good nor poor'] + df['% Fairly poor'], label='% Very poor', color='lightsteelblue')

# Add labels and title
plt.title('Percentage Distribution of Overall Experience by Region')
plt.xlabel('Region')
plt.ylabel('Percentage')
plt.xticks(rotation=90)
plt.legend()

# Save the chart as a PNG file
chart_filename = 'overall_experience_chart.png'
plt.savefig(chart_filename, format='png')
plt.close()

# Create a new Excel workbook and insert the chart image
wb = Workbook()
ws = wb.active
ws.title = 'Chart'

# Insert the saved chart image into the Excel file
img = Image(chart_filename)
ws.add_image(img, 'A1')

# Save the Excel workbook
excel_filename = 'overall_experience_chart.xlsx'
wb.save(excel_filename)

print(f"Chart successfully exported to {excel_filename}")


```
